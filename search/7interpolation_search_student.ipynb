Interpolation search ใช้กับข้อมูลที่เรียงลำดับแล้ว และพยายาม `เดาตำแหน่ง` ของค่าที่ต้องการหาโดยดูจากสัดส่วนของค่า ไม่ได้เลือกจุดกลางแบบ binary search เสมอไป


## แนวคิดหลัก

ถ้า `item` อยู่ใกล้ค่าด้านซ้ายของช่วง ก็มีแนวโน้มว่าตำแหน่งจริงจะอยู่ใกล้ฝั่งซ้าย

ถ้า `item` อยู่ใกล้ค่าด้านขวาของช่วง ก็มีแนวโน้มว่าตำแหน่งจริงจะอยู่ใกล้ฝั่งขวา

จึงคำนวณตำแหน่งประมาณด้วยสูตรนี้:

```python
point = left + ((item - arr[left]) * (right - left)) // (arr[right] - arr[left])
```

หลังจากนั้น:

1. ถ้าค่าที่ `point` เท่ากับ `item` ก็เจอคำตอบ
2. ถ้า `item` น้อยกว่า ก็ขยับช่วงไปทางซ้าย
3. ถ้า `item` มากกว่า ก็ขยับช่วงไปทางขวา

เพราะฉะนั้น interpolation search คล้าย binary search ตรงที่ค่อย ๆ ลดช่วงค้นหา แต่ต่างกันตรงที่มันไม่ได้ใช้จุดกึ่งกลางตายตัว

## ความซับซ้อน

| เรื่อง | ค่า |
|---|---|
| Average Time Complexity | `O(log log n)` เมื่อข้อมูลกระจายค่อนข้างสม่ำเสมอ |
| Worst-case Time Complexity | `O(n)` |
| Space Complexity แบบ loop | `O(1)` |
| Space Complexity แบบ recursion | ขึ้นกับความลึกของการเรียกซ้ำ |

ดังนั้น interpolation search อาจเร็วมากในข้อมูลบางแบบ แต่ถ้าข้อมูลกระจายไม่สม่ำเสมอ ข้อได้เปรียบนี้อาจหายไป

## กรณีพิเศษที่ต้องระวัง

ในสูตรของ interpolation search มีการหารด้วยค่า:

```python
arr[right] - arr[left]
```

ถ้าค่าทางซ้ายและขวาเท่ากัน เช่น `[5, 5, 5, 5, 5]` จะเกิดการหารด้วยศูนย์ได้

จึงต้องเช็กก่อนว่า:

- ถ้าค่าซ้ายและขวาเท่ากัน และเท่ากับ `item` ด้วย ก็คืนตำแหน่งนั้นได้เลย
- ถ้าไม่เท่ากับ `item` ก็แปลว่าไม่มีคำตอบในช่วงนั้น

## 1. interpolation_search

ฟังก์ชันนี้เป็นเวอร์ชันใช้ `while` loop

ถ้าเจอ `item` จะคืน index ถ้าไม่เจอจะคืน `None`

````python

def interpolation_search(sorted_collection: list[int], item: int,left: int = 0, right = None):
    # TODO:
    # left = 0
    right = len(sorted_collection) -1
    if  not sorted_collection:
        return None
    
    # while left <= right:
    point = left + ((item - sorted_collection[left])*(right - left))// (sorted_collection[right] - sorted_collection[left])
    print(f"คำนวณ point ได้ {point}")
    if point < 0 or point >= len(sorted_collection):
        print("point หลุดช่วงของ list จึงถือว่าไม่เจอ")
        return None
    
    current_item = sorted_collection[point]
    
    if current_item == item:
        print(f"เจอ {item} ที่ index {point}")
        return point
    if item < current_item:
        print("target น้อยกว่า current_item จึงตัดช่วงขวาทิ้ง")
        right = point - 1
        return interpolation_search(sorted_collection,item,left, right)
    else:
        print("target มากกว่า current_item จึงตัดช่วงซ้ายทิ้ง")
        left = point + 1
        return interpolation_search(sorted_collection,item,left, right)
    print("ไม่เจอ target")
    return None
    # 1. กำหนด left = 0 และ right = len(sorted_collection) - 1
    # 2. วนลูป while left <= right
    # 3. เช็กกรณีค่าซ้ายและขวาเท่ากัน เพื่อป้องกันการหารด้วยศูนย์
    # 4. คำนวณ point ด้วยสูตร interpolation search
    # 5. ถ้า point อยู่นอกช่วงของ list ให้ return None
    # 6. เปรียบเทียบค่าที่ point กับ item
    # 7. ถ้าเจอให้ return point ถ้าไม่เจอให้ปรับ left หรือ right
    # 8. ถ้าวนจนหมดแล้วยังไม่เจอ ให้ return None
    # pass
````
