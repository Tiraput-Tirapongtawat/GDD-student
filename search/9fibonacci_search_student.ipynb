3. ถ้าเจอค่าที่ต้องการ ให้คืน index
4. ถ้า target น้อยกว่าค่าที่ตรวจ ให้ลดช่วงไปทางซ้าย
5. ถ้า target มากกว่าค่าที่ตรวจ ให้เลื่อน `offset` แล้วค้นหาช่วงขวาต่อ

## ลำดับ Fibonacci

ลำดับ Fibonacci เริ่มจาก `0, 1` แล้วตัวถัดไปเกิดจากผลบวกของ 2 ตัวก่อนหน้า

```text
0, 1, 1, 2, 3, 5, 8, 13, 21, ...
```

ตัวอย่าง:

- `fibonacci(0)` ได้ `0`
- `fibonacci(1)` ได้ `1`
- `fibonacci(5)` ได้ `5`
- `fibonacci(7)` ได้ `13`

|---|---|
