## แนวคิดหลัก

แทนที่จะ binary search ทั้ง list ตั้งแต่ต้น exponential search จะทำ 2 ช่วง:

1. หา boundary ก่อน โดยเริ่มจาก index `1` แล้วขยายเป็น `2`, `4`, `8`, ... จนค่าที่ตำแหน่งนั้นมากกว่าหรือเท่ากับ `item`
2. เมื่อรู้ช่วงที่ `item` อาจอยู่แล้ว จึงใช้ binary search เฉพาะช่วงนั้น

เช่น ถ้าต้องหา `25` ใน list เรียงลำดับ อัลกอริทึมจะขยายช่วงไปเรื่อย ๆ จนรู้ว่า `25` น่าจะอยู่ระหว่าง index ใด แล้วค่อยค้นหาในช่วงนั้น

## ความซับซ้อน

| เรื่อง | ค่า |
|---|---|
| Time Complexity | `O(log i)` เมื่อ `i` คือ index ของ item ที่พบ |
| Worst-case Time Complexity | `O(log n)` |
| Space Complexity | `O(log n)` จาก recursion ของ binary search |

ข้อดีคือถ้า item อยู่ใกล้ต้น list มาก ๆ อาจเจอช่วงค้นหาเร็วกว่าเริ่ม binary search ทั้ง list ทันที
