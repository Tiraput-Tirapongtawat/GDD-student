````python
def binary_search(test, target):
    high = len(test)
    # midpoint = length_test //2
    
    low = 0
    
    while low <= high:
        mid_point = (low +high )//2
        current_test = test[mid_point]
        
        if target == current_test:
            return True, mid_point
        
        if target < current_test:
            high = mid_point -1
        else:
            low = mid_point + 1
    return False

print(binary_search(test_data,12))

        
````

````python
def binary_recursive_search(test, target):
    midpoint = len(test) // 2
    
    if len(test) == 0:
        return False
    
    if test[midpoint] == target:
        return True
    
    if test[midpoint] > target:
        return binary_recursive_search(test[:midpoint],target)
    
    return binary_recursive_search(test[midpoint + 1:], target)
    
    
    
````

## 1. bisect_left

`bisect_left` ใช้หาตำแหน่งที่ควรแทรก `item` เข้าไป โดยแทรก **ก่อนค่าที่เท่ากันตัวแรก**

พูดง่าย ๆ คือหาตำแหน่งแรกที่มีค่า `>= item`

ตัวอย่าง:

```python
[0, 5, 7, 10, 15]
```

ถ้าจะใส่ `6` ตำแหน่งที่เหมาะคือ index `2` เพราะควรอยู่ระหว่าง `5` กับ `7`

````python
def bisect_left(sorted_collection: list[int], item: int, lo: int = 0, hi: int = -1) -> int:
    # TODO:
    # 1. ถ้า hi < 0 ให้ตั้งเป็น len(sorted_collection)
    # 2. วน while ขณะที่ lo < hi
    # 3. หา mid และตัดช่วงแบบ binary search
    # 4. คืนตำแหน่งแรกที่มีค่า >= item
    if hi < 0 :
        hi = len(sorted_collection)
    while lo < hi:
        mid = lo + (hi - lo)// 2
        # print(mid)
        if sorted_collection[mid] < item:
            lo = mid + 1
        else:
            hi = mid
    return lo
    
````

````python
# รัน cell นี้หลังจากเขียน bisect_left เสร็จ
print(bisect_left([0, 5, 7, 10, 15], 0))
print(bisect_left([0, 5, 7, 10, 15], 6))
print(bisect_left([0, 5, 7, 10, 15], 20))
print(bisect_left([1, 2, 2, 2, 3], 2))
````
