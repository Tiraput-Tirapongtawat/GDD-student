## แนวคิดหลัก

ถ้า list มีขนาด `n` jump search จะเลือกขนาด block ประมาณ `sqrt(n)`

ขั้นตอนหลัก:

1. กระโดดไปทีละ block จนเจอค่าท้าย block ที่มากกว่าหรือเท่ากับ `item`
2. กลับมาค้นหาแบบ linear search ภายใน block นั้น
3. ถ้าเจอ ให้คืน index
4. ถ้าไม่เจอ ให้คืน `-1`

แนวคิดนี้อยู่กึ่งกลางระหว่าง linear search กับ binary search: ไม่ได้ตัดครึ่งข้อมูล แต่ก็ไม่ไล่ทีละตัวตั้งแต่ต้นเสมอไป

## ความซับซ้อน

| เรื่อง | ค่า |
|---|---|
| Time Complexity | `O(sqrt(n))` |
| Space Complexity | `O(1)` |

เงื่อนไขสำคัญคือข้อมูลต้องเรียงลำดับแล้ว เพราะการกระโดดข้าม block อาศัยการเปรียบเทียบว่าค่าที่เจอมากกว่าหรือน้อยกว่า `item`
