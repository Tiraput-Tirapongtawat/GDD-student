## แนวคิดหลัก

- `one_third` จุดประมาณหนึ่งในสามของช่วง
- `two_third` จุดประมาณสองในสามของช่วง

## ความซับซ้อน

| เรื่อง | ค่า |
|---|---|
| Time Complexity | `O(log3 n)` |
| Space Complexity แบบ loop | `O(1)` |
| Space Complexity แบบ recursion | `O(log3 n)` จาก call stack |

ในทางปฏิบัติ binary search มักนิยมกว่าในงานค้นหา list ธรรมดา เพราะเปรียบเทียบน้อยกว่าและโค้ดง่ายกว่า แต่ ternary search เป็นตัวอย่างที่ดีของการแบ่งช่วงมากกว่า 2 ส่วน
