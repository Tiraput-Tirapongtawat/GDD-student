## 8. การลบข้อมูลใน Dictionary

วิธีลบข้อมูลที่พบบ่อยมี 3 แบบ

- `pop(key)` ลบ key ที่ระบุ และคืน value ที่ถูกลบ
- `popitem()` ลบคู่ key-value ล่าสุด
- `del dict[key]` ลบ key ที่ระบุโดยไม่คืนค่า

```python
sample = {
    'key1': 'value1',
    'key2': 'value2',
    'key3': 'value3',
    'key4': 'value4',
}

# TODO: ลบ key1 ด้วย pop()
removed_value = sample.pop('key1')

# TODO: ลบ item ล่าสุดด้วย popitem()
last_item = sample.popitem()

# TODO: ลบ key2 ด้วย del
del sample["key2"]

print(removed_value)
print(last_item)
print("Remain  :",sample)
```

ถ้าลบ key ที่ไม่มีอยู่ด้วย `pop()` หรือ `del` จะเกิด `KeyError` ดังนั้นก่อนลบอาจตรวจด้วย `in` หรือใช้ `pop(key, default)` ได้

`items()` คืนข้อมูลเป็นคู่ `(key, value)` ทั้งหมด

## 10. clear() ล้างข้อมูลทั้งหมด

`clear()` ลบข้อมูลทั้งหมดใน dictionary แต่ตัวแปร dictionary ยังอยู่

`clear()` คืนค่า `None` เพราะหน้าที่ของมันคือแก้ dictionary เดิม

```python
dct = {'key1': 'value1', 'key2': 'value2'}
print('before:', dct)

# TODO: ล้างข้อมูลทั้งหมดใน dct
result = dct.clear()

print('clear returned:', result)
print('after:', dct)
```

## 12. copy() คัดลอก Dictionary

`copy()` ใช้คัดลอก dictionary เพื่อให้เราแก้ตัว copy ได้โดยไม่กระทบ dictionary เดิมในระดับชั้นนอก

แต่ `copy()` เป็น shallow copy ถ้า value ข้างในเป็น list หรือ dictionary ซ้อนกัน ส่วนที่ซ้อนอยู่ยังอาจอ้างถึง object เดิม

```python
original = {'name': 'Ada', 'age': 30}

# TODO: copy original ไปที่ copied
copied = original.copy()

# TODO: แก้ age ใน copied
copied['age'] = 12

print('original:', original)
print('copied:', copied)
```

## สรุปคำสั่งที่ใช้บ่อย

| คำสั่ง | ความหมาย |
|---|---|
| `dict[key]` | ดึงค่าด้วย key ถ้า key ไม่มีจะ error |
| `dict.get(key)` | ดึงค่าด้วย key ถ้า key ไม่มีจะได้ `None` หรือ default ที่กำหนด |
| `dict[key] = value` | เพิ่ม key ใหม่ หรือแก้ value ของ key เดิม |
| `key in dict` | ตรวจว่า key มีอยู่ใน dictionary หรือไม่ |
| `dict.pop(key)` | ลบ key ที่ระบุและคืน value ที่ถูกลบ |
| `dict.popitem()` | ลบคู่ key-value ล่าสุด |
| `del dict[key]` | ลบ key ที่ระบุ |
| `dict.items()` | คืนคู่ `(key, value)` ทั้งหมด |
| `dict.keys()` | คืน key ทั้งหมด |
| `dict.values()` | คืน value ทั้งหมด |
| `dict.clear()` | ล้างข้อมูลทั้งหมดใน dictionary |
| `dict.copy()` | copy dictionary แบบ shallow copy |
