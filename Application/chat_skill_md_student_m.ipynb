## 3) สร้าง client

สร้างตัวแปร `client` ด้วย `OpenAI(...)`

- `base_url` ใช้ค่าจาก `OPENAI_COMPAT_BASE`

````python
# TODO: สร้าง client สำหรับเชื่อมต่อ Ollama
from  openai import OpenAI

MODEL = "qwen3.5:4b"
OLLAMA_HOST = "http://127.0.0.1:11434"
OPENAI_COMPAT_BASE = f"{OLLAMA_HOST}/v1/"

client = OpenAI(
    base_url = OPENAI_COMPAT_BASE ,
    api_key="ollama",
)
````

เก็บผลไว้ในตัวแปร `response` แล้ว print คำตอบจาก:

````python
# TODO: ส่งคำถามแรกให้โมเดลและ print คำตอบ
response = client.chat.completions.create(
    model = MODEL,
    messages= [
        {"role":"system", "content": "คุณเป็นผู้ช่วยสอนที่เป็นมิตร ตอบไม่เกิน 2 ประโยค"},
        {"role":"user", "content": "ช่วยแนะนำ อาหารเช้าที่มีประโยชน์ครบ 5 หมู่ มา 5 เมนู"},
    ],
    reasoning_effort="none",
    max_tokens= 120
)
````

````python
chat_history = [
            {"role": "system", "content": "คุณเป็นผู้ช่วยสอนที่เป็นมิตร ตอบไม่เกิน 2 ประโยค"}]
````

````python
# TODO: สร้าง chat_history ที่มี system message หนึ่งรายการ
# TODO: เขียน def ask_bot(question) ให้สมบูรณ์
def chat_h(message):
    chat_history.append({"role": "user", "content":message})
    response = client.chat.completions.create(
        model=MODEL,
        messages=chat_history,
        reasoning_effort="none",
        max_tokens=120,
        temperature= 0.9
        
    )

    answer = response.choices[0].message.content
    chat_history.append({"role":"asistance", "content":answer})
    return answer
````
