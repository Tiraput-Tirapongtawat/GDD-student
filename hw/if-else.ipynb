## ข้อ 1: ประตูร้านราเมงอัจฉริยะ

ร้านราเมงมีระบบให้คิวอัตโนมัติจากจำนวนลูกค้าและสถานะสมาชิก

- ถ้า `customers` มากกว่า 8 คน และ `is_member` เป็น `True` ให้พิมพ์ `รับบัตรคิวด่วน`
- ถ้า `customers` มากกว่า 8 คน แต่ไม่ใช่สมาชิก ให้พิมพ์ `รอคิวปกติ`
- ถ้า `customers` อยู่ระหว่าง 1 ถึง 8 ให้พิมพ์ `เข้าร้านได้`
- ถ้า `customers` เท่ากับ 0 ให้พิมพ์ `ร้านว่าง`

```python
customers = 12
is_member = True

if customers > 8:
    if is_member:
        print("รับบัตรคิวด่วน")
    else:
        print("รอคิวปกติ")
elif 1 <= customers <= 8:
    print("เข้าร้านได้")
else:
    print("ร้านว่าง")
```

## ข้อ 26 [ท้าทาย]: ระบบสรุปความเสี่ยงงานเทศกาลกลางคืน

ผู้จัดงานต้องประเมินจาก `crowd`, `rain_chance`, `security_staff`, `medical_team_ready`, และ `stage_power`

ให้สร้างตัวแปร `risk_score` เริ่มที่ 0 แล้วเพิ่มคะแนนตามนี้ก่อนตัดสินผล:

- crowd มากกว่า 1000 เพิ่ม 3 คะแนน, ถ้า crowd มากกว่า 500 เพิ่ม 2 คะแนน, ถ้าน้อยกว่าหรือเท่ากับ 500 เพิ่ม 1 คะแนน
- rain_chance มากกว่า 70 เพิ่ม 2 คะแนน, ถ้า rain_chance ตั้งแต่ 40 ถึง 70 เพิ่ม 1 คะแนน
- security_staff น้อยกว่า 10 เพิ่ม 2 คะแนน
- medical_team_ready เป็น `False` เพิ่ม 3 คะแนน
- stage_power เป็น `unstable` เพิ่ม 2 คะแนน

หลังจากคำนวณ risk_score แล้วให้ตัดสิน:

- ถ้า medical_team_ready เป็น `False` และ crowd มากกว่า 1000 ให้พิมพ์ `เลื่อนงาน: ทีมแพทย์ไม่พร้อมกับผู้ชมจำนวนมาก`
- ถ้า stage_power เป็น `offline` ให้พิมพ์ `เลื่อนงาน: ระบบไฟเวทีไม่พร้อม`
- ถ้า risk_score ตั้งแต่ 8 ขึ้นไป ให้พิมพ์ `ความเสี่ยงสูง: ต้องประชุมฉุกเฉิน`
- ถ้า risk_score ตั้งแต่ 5 ถึง 7 ให้พิมพ์ `ความเสี่ยงกลาง: เพิ่มมาตรการความปลอดภัย`
- ถ้า risk_score น้อยกว่า 5 ให้พิมพ์ `จัดงานได้ตามแผน`

ข้อนี้ฝึกทั้งการสะสมคะแนนและการมีเงื่อนไข override ที่ต้องชนะ risk_score

```python
crowds = 1200
rain_chance = 70
security_staff = 8
medical_team_ready = True
stage_power = "unstable"
risk_score = 0

if crowds > 1000:
    risk_score += 3
elif crowds > 500 :
    risk_score += 2
else:
    risk_score += 1

if rain_chance > 70:
    risk_score += 2
elif rain_chance >= 40:
    risk_score += 1
    
if security_staff > 10:
    risk_score += 2
    
if not medical_team_ready:
    risk_score += 3
if stage_power == "unstable":
    risk_score += 2
    
    
if not medical_team_ready and crowds > 1000:
    print("เลื่อนงาน....")
elif stage_power == "offline":
    print("เลื่อนงาน....ระบบ")
elif risk_score >= 8 :
    print("ความเสี่ยงสูง....")
elif risk_score >= 5 :
    print("ความเสี่ยงกลาง...")
else:
    print("จัดงานได้ตามแผน")
```
