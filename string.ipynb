```python
a = 3
b = 5

print(a+b) # 3 + 5 = 8
print(a-b) # 3 - 5 = -2
print(a*b) # 3*5 = 15
print(a/b) # 3/5 = 0.6


# ans
print(f"{a} + {b} = {a + b}")
print(f"{a} - {b} = {a - b}")
print(f"{a} * {b} = {a * b}")
print(f"{a} / {b} = {a / b}")
```

# String Slicing

```python
# start stop step

string = "abcde"
print(len(string))
print(string[0])
print(string[1])
print(string[2])
print(string[3])
print(string[4])
print(string[5])
```

```python
string = "sdlglhbsldjgbsjhdfbgkjshebfgjsbfgkjhb"
string_num = len(string) -1
print(string[string_num])
```

```python
string = "sdlglhbsldjgbsjhdfbgkjshebfgjsbfgkjhb"
print(string[-1])
print(string[-2])
print(string[-3])
print(string[-4])
print(string[-6])
print(string[-9])
print(string[-10])
print(string[-20])
print(string[-100])
```

```python
string = "Lebron James"

# firstname = string[0:6]
firstname = string[:6]
print(firstname)
# lastname = string[7:12]
lastname = string[7:]
print(lastname)
```

```python
string = "12345678"

ood_num = string[::2] # start stop step
print(ood_num)
```

```python
string = "elephant"

invert_string = string[::-1]
print(invert_string)
```

```python
word = "elephant"
#aeiou

# Vowel : e
# Vowel : a
if "a" in word:
    print("Vowel a in word")
if "e" in word:
    print("Vowel e in word")
if "i" in word:
    print("Vowel i in word")
if "u" in word:
    print("Vowel u in word")
if "o" in word:
    print("Vowel o in word")
```

```python
for i in word:
    # print(i)
    if i in "aeiou":
        print(f"Vowel : {i}")
```
