### 🛟 Recovery Cell

รัน cell ด้านล่างได้เมื่อฟังก์ชันจาก Level ก่อนหน้าหายหรือทำงานไม่ถูกต้อง จากนั้นจึงทำโจทย์ของ Checkpoint ต่อ


````python
def normalize_text(text):
    return text.strip().title()


def create_pilot(name, callsign):
    return {
        "name": normalize_text(name),
        "callsign": normalize_text(callsign),
    }


def create_ship(name, fuel, capacity):
    if fuel < 0:
        fuel = 0
    return {
        "name": normalize_text(name),
        "fuel": fuel,
        "capacity": capacity,
    }


def create_planet(name, distance, is_open):
    return {
        "name": normalize_text(name),
        "distance": distance,
        "is_open": is_open,
    }


def create_mission(pilot, ship, planet, budget):
    return {
        "pilot": pilot,
        "ship": ship,
        "planet": planet,
        "budget": budget,
        "packages": [],
        "launched": False,
    }


def calculate_required_fuel(mission):
    distance = mission["planet"]["distance"]
    return (distance + 9) // 10


def create_package(name, weight, package_type):
    return {
        "name": normalize_text(name),
        "weight": weight,
        "type": normalize_text(package_type),
        "status": "รอตรวจสอบ",
    }


def calculate_package_cost(package):
    cost = package["weight"] * 15
    if package["type"] == "Fragile":
        cost += 20
    return cost


def total_cargo_weight(mission):
    total = 0
    for package in mission["packages"]:
        total += package["weight"]
    return total


def is_package_allowed(mission, package):
    weight_ok = package["weight"] <= 20
    type_ok = package["type"] not in ["Weapon", "Explosive"]
    capacity_ok = (
        total_cargo_weight(mission) + package["weight"]
        <= mission["ship"]["capacity"]
    )
    return weight_ok and type_ok and capacity_ok


def add_package(mission, package):
    if is_package_allowed(mission, package):
        package["status"] = "พร้อมส่ง"
        mission["packages"].append(package)
        return True
    return False


def total_shipping_cost(mission):
    total = 0
    for package in mission["packages"]:
        total += calculate_package_cost(package)
    return total


def find_package(mission, package_name):
    target = normalize_text(package_name)
    for package in mission["packages"]:
        if package["name"] == target:
            return package
    return None


def update_package_status(mission, package_name, new_status):
    package = find_package(mission, package_name)
    if package is None:
        return False
    package["status"] = new_status
    return True
````

### 🛟 Recovery Cell

รัน cell ด้านล่างได้เมื่อฟังก์ชันจาก Level ก่อนหน้าหายหรือทำงานไม่ถูกต้อง จากนั้นจึงทำโจทย์ของ Checkpoint ต่อ


````python
def normalize_text(text):
    return text.strip().title()


def create_pilot(name, callsign):
    return {
        "name": normalize_text(name),
        "callsign": normalize_text(callsign),
    }


def create_ship(name, fuel, capacity):
    if fuel < 0:
        fuel = 0
    return {
        "name": normalize_text(name),
        "fuel": fuel,
        "capacity": capacity,
    }


def create_planet(name, distance, is_open):
    return {
        "name": normalize_text(name),
        "distance": distance,
        "is_open": is_open,
    }


def create_mission(pilot, ship, planet, budget):
    return {
        "pilot": pilot,
        "ship": ship,
        "planet": planet,
        "budget": budget,
        "packages": [],
        "launched": False,
    }


def calculate_required_fuel(mission):
    distance = mission["planet"]["distance"]
    return (distance + 9) // 10


def create_package(name, weight, package_type):
    return {
        "name": normalize_text(name),
        "weight": weight,
        "type": normalize_text(package_type),
        "status": "รอตรวจสอบ",
    }


def calculate_package_cost(package):
    cost = package["weight"] * 15
    if package["type"] == "Fragile":
        cost += 20
    return cost


def total_cargo_weight(mission):
    total = 0
    for package in mission["packages"]:
        total += package["weight"]
    return total


def is_package_allowed(mission, package):
    weight_ok = package["weight"] <= 20
    type_ok = package["type"] not in ["Weapon", "Explosive"]
    capacity_ok = (
        total_cargo_weight(mission) + package["weight"]
        <= mission["ship"]["capacity"]
    )
    return weight_ok and type_ok and capacity_ok


def add_package(mission, package):
    if is_package_allowed(mission, package):
        package["status"] = "พร้อมส่ง"
        mission["packages"].append(package)
        return True
    return False


def total_shipping_cost(mission):
    total = 0
    for package in mission["packages"]:
        total += calculate_package_cost(package)
    return total


def find_package(mission, package_name):
    target = normalize_text(package_name)
    for package in mission["packages"]:
        if package["name"] == target:
            return package
    return None


def update_package_status(mission, package_name, new_status):
    package = find_package(mission, package_name)
    if package is None:
        return False
    package["status"] = new_status
    return True


def cargo_summary(mission):
    return {
        "count": len(mission["packages"]),
        "weight": total_cargo_weight(mission),
        "cost": total_shipping_cost(mission),
    }


def packages_for_destination(mission, destination):
    target = normalize_text(destination)
    selected = []
    for package in mission["packages"]:
        if package.get("destination") == target:
            selected.append(package)
    return selected


def fuel_is_ready(mission):
    required = calculate_required_fuel(mission)
    return mission["ship"]["fuel"] >= required


def budget_is_ready(mission):
    return mission["budget"] >= total_shipping_cost(mission)


def cargo_is_ready(mission):
    has_packages = len(mission["packages"]) > 0
    within_capacity = (
        total_cargo_weight(mission) <= mission["ship"]["capacity"]
    )
    return has_packages and within_capacity


def check_mission(mission):
    reasons = []
    if not mission["planet"]["is_open"]:
        reasons.append("สถานีปลายทางปิด")
    if not fuel_is_ready(mission):
        reasons.append("เชื้อเพลิงไม่เพียงพอ")
    if not budget_is_ready(mission):
        reasons.append("งบประมาณไม่เพียงพอ")
    if not cargo_is_ready(mission):
        reasons.append("ห้องเก็บของยังไม่พร้อม")
    return {"ready": len(reasons) == 0, "reasons": reasons}


def show_manifest(mission):
    if len(mission["packages"]) == 0:
        print("ยังไม่มีพัสดุ")
        return
    number = 1
    for package in mission["packages"]:
        cost = calculate_package_cost(package)
        print(
            f"{number}. {package['name']} | {package['weight']} กก. | "
            f"{package['status']} | {cost} เครดิต"
        )
        number += 1


def show_readiness(mission):
    summary = cargo_summary(mission)
    result = check_mission(mission)
    print(f"พัสดุ: {summary['count']} กล่อง")
    print(f"น้ำหนักรวม: {summary['weight']} กิโลกรัม")
    print(f"ค่าส่งรวม: {summary['cost']} เครดิต")
    if result["ready"]:
        print("สถานะ: พร้อมออกเดินทาง")
    else:
        print("สถานะ: ยังไม่พร้อม")
        for reason in result["reasons"]:
            print(f"- {reason}")


def input_package():
    name = input("ชื่อพัสดุ: ")
    weight = float(input("น้ำหนัก (กก.): "))
    package_type = input("ประเภทพัสดุ: ")
    return create_package(name, weight, package_type)


def add_package_from_input(mission):
    package = input_package()
    if add_package(mission, package):
        print(f"เพิ่ม {package['name']} สำเร็จ")
    else:
        print(f"ไม่สามารถเพิ่ม {package['name']} ได้")


def show_menu():
    print("=== เมนูควบคุมภารกิจ ===")
    print("1. เพิ่มพัสดุ")
    print("2. ดูบัญชีพัสดุ")
    print("3. ตรวจความพร้อม")
    print("4. ออกเดินทาง")
    print("5. ปิดระบบ")


def launch_mission(mission):
    result = check_mission(mission)
    if not result["ready"]:
        print("ไม่สามารถออกเดินทางได้")
        for reason in result["reasons"]:
            print(f"- {reason}")
        return False
    mission["ship"]["fuel"] -= calculate_required_fuel(mission)
    for package in mission["packages"]:
        package["status"] = "กำลังจัดส่ง"
    mission["launched"] = True
    print("ยานออกเดินทางแล้ว!")
    return True


def show_mission_report(mission):
    summary = cargo_summary(mission)
    status = "ออกเดินทางแล้ว" if mission["launched"] else "ยังไม่ออกเดินทาง"
    print("=== รายงานภารกิจ ===")
    print(f"นักบิน: {mission['pilot']['name']}")
    print(f"ยาน: {mission['ship']['name']}")
    print(f"ปลายทาง: {mission['planet']['name']}")
    print(f"พัสดุ: {summary['count']} กล่อง")
    print(f"น้ำหนัก: {summary['weight']} กิโลกรัม")
    print(f"สถานะ: {status}")


def handle_choice(choice, mission):
    if choice == "1":
        add_package_from_input(mission)
    elif choice == "2":
        show_manifest(mission)
    elif choice == "3":
        show_readiness(mission)
    elif choice == "4":
        launch_mission(mission)
    elif choice == "5":
        print("ปิดระบบควบคุม")
        return False
    else:
        print("ไม่มีเมนูนี้")
    return True


def run_delivery_system(mission):
    running = True
    while running:
        show_menu()
        choice = input("เลือกเมนู: ")
        running = handle_choice(choice, mission)
````
